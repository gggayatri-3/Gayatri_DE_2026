Seaborn Notes

Seaborn is a Python library built on top of Matplotlib. It is designed for statistical data visualization and makes it easier to create attractive and readable plots with less code.

What Seaborn is used for

Exploratory data analysis
Distribution and relationship plots

Statistical comparisons

Clean and modern-looking charts
Why it matters

Seaborn is especially useful when working with pandas DataFrames. It provides built-in themes and functions that help create polished visuals quickly.

1. Import Required Libraries

Seaborn is commonly used with pandas and Matplotlib. These libraries help create clean and informative visualizations easily.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

print('Seaborn is ready!')
```

Notes

seaborn provides statistical plotting functions.
matplotlib handles the final rendering of the chart.
pandas is useful for organizing and exploring tabular data

2. Load and Explore Dataset

We can create a small dataset of monthly sales for two products.

```python
sales_df = pd.DataFrame({
    'month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'sales': [120, 150, 170, 165, 210, 240],
    'product': ['A', 'A', 'A', 'B', 'B', 'B']
})

print(sales_df)
print('\nShape:', sales_df.shape)
print('\nFirst rows:')
print(sales_df.head())
print('\nSummary:')
print(sales_df.describe())
```

Notes

Data exploration helps us understand values before plotting.
It is important to know whether the data is categorical, numerical, or time-based.
Seaborn works very well with pandas DataFrames.

3. Create a Line Plot

Line plots help visualize change over time.

```python
sns.set_style('whitegrid')
plt.figure(figsize=(8, 5))
sns.lineplot(data=sales_df, x='month', y='sales', marker='o')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()
```

🔑 Key Insights from the Visualization

Trend analysis: The line plot highlights whether sales are increasing, decreasing, or fluctuating across months.

Seasonality detection: Peaks and dips can reveal seasonal patterns (e.g., higher sales in festive months).

Performance monitoring: Helps track monthly performance and identify strong or weak months.

Decision support: Useful for business decisions like inventory planning, marketing campaigns, or resource allocation.

Notes

sns.lineplot() creates a clean line chart with a polished default style.

It is useful for trends over time.

This chart makes it easy to see whether values are increasing or decreasing.

4. Create a Bar Chart

Bar charts are useful for comparing categories.

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=sales_df, x='month', y='sales', palette='Set2')
plt.title('Monthly Sales by Month')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()
```

🔑 Key Insights from the Visualization

Comparison across months: The bar plot emphasizes differences in sales between months more clearly than a line plot.

Top and bottom performers: Easy to spot which months had the highest and lowest sales.

Distribution clarity: Highlights relative proportions of sales across months, useful for identifying imbalances.

Business decisions: Helps in resource allocation, marketing focus, and inventory planning based on strong vs weak months.

Notes

sns.barplot() makes it easy to compare different groups.

Bar charts are commonly used for category totals and counts.

Seaborn adds good default styling automatically.

5. Create a Histogram

A histogram shows the distribution of one numerical feature.

```python
scores = [55, 60, 62, 65, 68, 70, 72, 75, 78, 80, 82, 85, 90, 92]

plt.figure(figsize=(8, 5))
sns.histplot(scores, bins=6, kde=True, color='steelblue')
plt.title('Distribution of Scores')
plt.xlabel('Score')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

🔑 Key Insights from the Visualization

Central tendency: The KDE curve helps identify where most scores cluster (around 70–85).

Spread of scores: Shows how scores range from 55 to 92, with most values concentrated in the higher bins.

Performance grouping: Easy to see how many students fall into low, medium, and high score ranges.

Shape of distribution: The histogram suggests a right-skewed distribution, with more students scoring higher.

Notes

Histogram bins decide how values are grouped.
This chart helps identify center, spread, and skewness.
It is a useful way to inspect data distribution before analysis.

6. Create a Scatter Plot

Scatter plots help examine the relationship between two numeric variables.

```python
import numpy as np

x = np.array([1, 2, 3, 4, 5, 6, 7])
y = np.array([2, 5, 6, 8, 10, 12, 14])

plt.figure(figsize=(8, 5))
sns.scatterplot(x=x, y=y, color='darkgreen', s=80)
plt.title('Relationship Between Variables')
plt.xlabel('X Variable')
plt.ylabel('Y Variable')
plt.tight_layout()
plt.show()
```

🔑 Key Insights from the Visualization

Positive correlation: As x increases, y also increases, suggesting a strong positive relationship.

Linear trend: The points roughly align along a straight line, indicating a near-linear relationship.

Predictive potential: This relationship could be modeled with a regression line to predict y values from x.

Data clarity: Scatter plots are excellent for spotting outliers, but here the data appears consistent without anomalies.

Notes

Scatter plots show whether two variables move together.

If points rise from left to right, the relationship may be positive.

This is useful for exploring correlation and clustering.

7. Create a Box Plot

Box plots show how values are distributed across categories and help detect outliers.

```python
region_data = pd.DataFrame({
    'region': ['North', 'North', 'North', 'South', 'South', 'South', 'East', 'East', 'East'],
    'sales': [120, 150, 170, 100, 130, 140, 160, 180, 200]
})

plt.figure(figsize=(8, 5))
sns.boxplot(data=region_data, x='region', y='sales', palette='pastel')
plt.title('Sales Distribution by Region')
plt.xlabel('Region')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()
```

🔑 Key Insights from the Visualization

Median sales: The line inside each box shows the median sales for that region, useful for comparing central tendencies.

Spread of sales: The box height indicates variability; wider boxes mean more fluctuation in sales.

Outliers detection: Any points outside the whiskers highlight unusual sales values.

Regional comparison: Easy to see which region generally performs better (e.g., East appears to have higher sales overall compared to North and South).

Business implications: Identifying regions with consistent vs volatile sales helps in resource allocation and strategy planning.

Notes

The box shows the middle 50% of the data.
The median line shows the midpoint.
Outliers appear as separate points outside the whiskers.

8. Create a Heatmap

A heatmap shows values across rows and columns using color intensity.

```python
matrix = pd.DataFrame(
    [[1, 2, 3], [2, 4, 5], [3, 5, 6]],
    index=['A', 'B', 'C'],
    columns=['X', 'Y', 'Z']
)

plt.figure(figsize=(7, 5))
sns.heatmap(matrix, annot=True, cmap='YlGnBu')
plt.title('Example Heatmap')
plt.tight_layout()
plt.show()
```

🔑 Key Insights from the Visualization

Value intensity: Darker cells represent higher values (e.g., 6 in row C, column Z), while lighter cells represent lower values.

Pattern recognition: The matrix shows increasing values across rows and columns, suggesting a structured progression.

Comparison across regions: Easy to compare how values differ between rows (A, B, C) and columns (X, Y, Z).

Data interpretation: Useful for spotting correlations, clusters, or anomalies in tabular data.

Notes

Heatmaps are useful for comparing many values at once.
Darker or lighter colors show differences in intensity.
They are often used for correlation tables and matrix-style summaries.

9. Style and Annotate Plots

Seaborn includes beautiful default themes, but you can also customize the visual style.

```python
sns.set_theme(style='whitegrid')
plt.figure(figsize=(8, 5))
sns.lineplot(data=sales_df, x='month', y='sales', marker='o', color='darkorange')
plt.title('Styled Sales Chart')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True, alpha=0.4)
plt.tight_layout()
plt.show()
```

Key Insights from the Visualization

Sales trend: The line chart highlights whether sales are rising, falling, or fluctuating across months.

Data clarity: Markers make each month’s sales value easy to identify, while the grid supports precise reading.

Visual emphasis: The orange line stands out against the white grid, drawing attention to the trend.

Business interpretation: Useful for identifying growth phases, seasonal dips, or consistent performance.

Notes

Style choices affect readability and clarity.
Good colors and spacing make charts easier to understand.
A simple, clean design often works best for presentations and reports.

10. Save and Export Figures

You can save a Seaborn plot as an image for use in reports or slides.

```python
plt.figure(figsize=(8, 5))
sns.lineplot(data=sales_df, x='month', y='sales', marker='o', color='seagreen')
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.tight_layout()
plt.savefig('seaborn_sales.png', dpi=300)
plt.show()
print('Chart saved as seaborn_sales.png')
```

Key Insights from the Visualization

Sales trend: The chart highlights how sales evolve month by month, showing growth, decline, or fluctuations.

Data clarity: Markers make each month’s sales value easy to identify.

Professional output: Saving at 300 dpi ensures the chart is publication-quality, ideal for business documents.

Reusability: The saved file can be reused in presentations, dashboards, or shared with stakeholders.

Many Graph Examples Using the Seaborn Tips Dataset

All examples below use the tips dataset from Seaborn, which is perfect for learning data visualization.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

sns.set_theme(style='whitegrid')
tips = sns.load_dataset('tips')
print(tips.head())

# 1. Line plot: average bill by day
plt.figure(figsize=(8, 4))
sns.lineplot(data=tips, x='day', y='total_bill', estimator='mean', marker='o')
plt.title('1. Line Plot: Average Bill by Day')
plt.xlabel('Day')
plt.ylabel('Average Total Bill')
plt.show()

# 2. Bar plot: average tip by day
plt.figure(figsize=(8, 4))
sns.barplot(data=tips, x='day', y='tip', estimator='mean', palette='Set2')
plt.title('2. Bar Plot: Average Tip by Day')
plt.xlabel('Day')
plt.ylabel('Average Tip')
plt.show()

# 3. Bar plot: total bill by sex
plt.figure(figsize=(7, 4))
sns.barplot(data=tips, x='sex', y='total_bill', estimator='mean', palette='Blues')
plt.title('3. Bar Plot: Average Bill by Sex')
plt.xlabel('Sex')
plt.ylabel('Average Total Bill')
plt.show()

# 4. Histogram: total bill distribution
plt.figure(figsize=(8, 4))
sns.histplot(tips['total_bill'], bins=20, kde=True, color='darkblue')
plt.title('4. Histogram: Total Bill Distribution')
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.show()

# 5. Scatter plot: total bill vs tip
plt.figure(figsize=(8, 4))
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex', alpha=0.8)
plt.title('5. Scatter Plot: Total Bill vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()

# 6. Scatter plot: bill vs size
plt.figure(figsize=(8, 4))
sns.scatterplot(data=tips, x='total_bill', y='size', hue='day', alpha=0.8)
plt.title('6. Scatter Plot: Bill vs Party Size')
plt.xlabel('Total Bill')
plt.ylabel('Party Size')
plt.show()

# 7. Box plot: total bill by day
plt.figure(figsize=(8, 4))
sns.boxplot(data=tips, x='day', y='total_bill')
plt.title('7. Box Plot: Total Bill by Day')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()

# 8. Box plot: tip by time
plt.figure(figsize=(7, 4))
sns.boxplot(data=tips, x='time', y='tip')
plt.title('8. Box Plot: Tip by Meal Time')
plt.xlabel('Time')
plt.ylabel('Tip')
plt.show()

# 9. Violin plot: total bill by smoker status
plt.figure(figsize=(8, 4))
sns.violinplot(data=tips, x='smoker', y='total_bill')
plt.title('9. Violin Plot: Total Bill by Smoker Status')
plt.xlabel('Smoker')
plt.ylabel('Total Bill')
plt.show()

# 10. Count plot: counts by day
plt.figure(figsize=(8, 4))
sns.countplot(data=tips, x='day', order=['Thur', 'Fri', 'Sat', 'Sun'])
plt.title('10. Count Plot: Observations by Day')
plt.xlabel('Day')
plt.ylabel('Count')
plt.show()

# 11. Pie chart using counts
label_counts = tips['day'].value_counts()
plt.figure(figsize=(7, 7))
plt.pie(label_counts.values, labels=label_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('11. Pie Chart: Day Share')
plt.show()

# 12. Heatmap: correlation matrix
corr = tips[['total_bill', 'tip', 'size']].corr()
plt.figure(figsize=(6, 5))
sns.heatmap(corr, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('12. Heatmap: Correlation Matrix')
plt.show()
```

🔑 Insights from Each Visualization

Line plot: Average bill by day  
Shows how average total bills vary across weekdays. Useful for spotting which days generate higher revenue.

Bar plot: Average tip by day  
Highlights tipping behavior by day. Can reveal patterns like higher tips on weekends.

Bar plot: Average bill by sex  
Compares spending between male and female customers. Useful for demographic insights.

Histogram: Total bill distribution  
Shows the spread of total bills, with KDE curve revealing central tendency and skewness.

Scatter plot: Total bill vs tip  
Displays correlation between bill size and tip amount, with color-coded sex differences.

Scatter plot: Bill vs party size  
Examines how party size influences total bill, with day-based color coding.

Box plot: Total bill by day  
Summarizes distribution of bills per day, showing medians, variability, and outliers.

Box plot: Tip by time  
Compares tipping behavior between lunch and dinner.

Violin plot: Total bill by smoker status  
Shows distribution shape and spread of bills for smokers vs non-smokers.

Count plot: Observations by day  
Displays frequency of records per day, useful for dataset balance.

Pie chart: Day share  
Visualizes proportion of observations across days.

Heatmap: Correlation matrix  
Shows correlations between total_bill, tip, and size. Reveals strong positive correlation between bill and tip.

Notes

Seaborn makes statistical plots look clean and readable.

The tips dataset is excellent for learning how to compare categories and distributions.

A well-designed plot helps answer a question clearly and quickly.
