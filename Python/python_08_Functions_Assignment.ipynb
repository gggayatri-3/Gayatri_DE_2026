```python
#Assignment 12: Check Leap Year
# Function to check whether a year is a leap year
def is_leap_year(year):
    if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
        return True
    else:
        return False

# Accept year from the user
year = int(input("Enter a year: "))

# Call the function and print the result
if is_leap_year(year):
    print(year, "is a leap year.")
else:
    print(year, "is not a leap year.")
```

```python
#Assignment 13: Factorial Using a Function
# Function to calculate factorial
def calculate_factorial(num):
    if num < 0:
        return None

    factorial = 1
    for i in range(1, num + 1):
        factorial *= i

    return factorial

# Accept input from the user
number = int(input("Enter a non-negative integer: "))

# Call the function and display the result
result = calculate_factorial(number)

if result is None:
    print("Factorial is not defined for negative numbers.")
else:
    print("Factorial of", number, "is", result)
```

```python
#Assignment 17: Palindrome Checker
# Function to check whether a string is a palindrome
def is_palindrome(text):
    # Remove spaces and convert to lowercase

    text = text.replace(" ", "").lower()

    # Check if the string is equal to its reverse

    return text == text[::-1]


# Accept input from the user

string = input("Enter a string: ")

# Call the function and print the result

if is_palindrome(string):
    print("The string is a palindrome.")
else:
    print("The string is not a palindrome.")
```

```python
#Assignment 19: Find Maximum in a List
# Function to find the maximum element in a list
def find_list_maximum(numbers):
    # Handle an empty list
    if len(numbers) == 0:
        return "The list is empty."

    # Assume the first element is the largest
    maximum = numbers[0]

    # Compare each element with the current maximum
    for num in numbers:
        if num > maximum:
            maximum = num

    return maximum

# Accept input from the user
numbers = list(map(int, input("Enter numbers separated by spaces: ").split()))

# Call the function
result = find_list_maximum(numbers)

# Display the result
print("Result:", result)
```

```python
#Assignment 20: Count Positive, Negative and Zero Values
# Function to count positive, negative, and zero values

def count_number_types(numbers):
    positive = 0
    negative = 0
    zero = 0

    for num in numbers:
        if num > 0:
            positive += 1
        elif num < 0:
            negative += 1
        else:
            zero += 1

    return positive, negative, zero

# Accept input from the user
numbers = list(map(int, input("Enter integers separated by spaces: ").split()))

# Call the function using tuple unpacking
positive_count, negative_count, zero_count = count_number_types(numbers)

# Display the results
print("Positive numbers:", positive_count)
print("Negative numbers:", negative_count)
print("Zero values:", zero_count)
```

```python
#Assignment 30: Mixed Parameter Types
# Function definition
def generate_report(report_id, /, title, author="Unknown", *scores, status="Draft", **details):
    print("\n----- Report -----")
    print("Report ID:", report_id)          # Positional-only parameter
    print("Title:", title)                  # Normal parameter
    print("Author:", author)                # Default parameter

    print("\nScores:")
    if scores:
        for score in scores:
            print(score)
    else:
        print("No scores available.")

    print("\nStatus:", status)              # Keyword-only parameter

    print("\nAdditional Details:")
    if details:
        for key, value in details.items():
            print(f"{key}: {value}")
    else:
        print("No additional details.")

# Accept input from the user
report_id = int(input("Enter Report ID: "))
title = input("Enter Report Title: ")
author = input("Enter Author Name: ")

num_scores = int(input("Enter number of scores: "))
scores = []

for i in range(num_scores):
    score = int(input(f"Enter score {i + 1}: "))
    scores.append(score)

status = input("Enter Report Status: ")

department = input("Enter Department: ")
year = input("Enter Year: ")

# Function call
generate_report(
    report_id,
    title,
    author,
    *scores,
    status=status,
    department=department,
    year=year
)
```

```python
#Assignment 31: Return Multiple Values
# Function definition
def number_statistics(numbers):
    minimum = min(numbers)
    maximum = max(numbers)
    total = sum(numbers)
    average = total / len(numbers)
    return minimum, maximum, total, average

# Accept input from the user
count = int(input("Enter how many numbers: "))

numbers = []
for i in range(count):
    num = float(input(f"Enter number {i + 1}: "))
    numbers.append(num)

# Tuple unpacking
minimum, maximum, total, average = number_statistics(numbers)

# Display the results
print("\nResults")
print("Minimum:", minimum)
print("Maximum:", maximum)
print("Total:", total)
print("Average:", average)
```

```python
#Assignment 32: Function Returning a Dictionary.
# Function definition
def analyze_text(text):
    vowels = "aeiouAEIOU"

    result = {
        "Total Characters": len(text),
        "Total Words": len(text.split()),
        "Total Vowels": 0,
        "Total Digits": 0,
        "Total Spaces": 0
    }

    for char in text:
        if char in vowels:
            result["Total Vowels"] += 1
        if char.isdigit():
            result["Total Digits"] += 1
        if char == " ":
            result["Total Spaces"] += 1

    return result

# Accept input from the user
text = input("Enter a string: ")

# Call the function
analysis = analyze_text(text)

# Display the results
print("\nText Analysis")
for key, value in analysis.items():
    print(f"{key}: {value}")
```

```python
#Assignment 33: Function Returning a List
# Function definition
def get_even_numbers(start, end):
    if start >= end:
        return None

    even_numbers = []

    for num in range(start, end + 1):
        if num % 2 == 0:
            even_numbers.append(num)

    return even_numbers

# Accept input from the user
start = int(input("Enter the starting number: "))
end = int(input("Enter the ending number: "))

# Validate and call the function
result = get_even_numbers(start, end)

if result is None:
    print("Error: Starting number must be smaller than the ending number.")
else:
    print("\nEven numbers are:")
    print(result)
```

```python
#Assignment 34: Prime Number Function
# Function definition
def is_prime(number):
    if number <= 1:
        return False

    for i in range(2, number):
        if number % i == 0:
            return False

    return True

# Print all prime numbers from 1 to 100
print("Prime numbers from 1 to 100 are:")

for num in range(1, 111):
    if is_prime(num):
        print(num, end=" ")
```

```python
#Assignment 35: Fibonacci Function
# Function definition
def generate_fibonacci(terms):
    if terms < 0:
        return "Error: Number of terms cannot be negative."
    elif terms == 0:
        return []

    fibonacci = []

    a, b = 0, 1

    for i in range(terms):
        fibonacci.append(a)
        a, b = b, a + b

    return fibonacci


# Accept input from the user
terms = int(input("Enter the number of terms: "))

# Call the function
result = generate_fibonacci(terms)

# Display the result
print("Fibonacci Series:", result)
```

```python
#Assignment 90: Closure-Based Discount Calculator
# Outer function
def create_discount(discount_percentage):

    # Inner function
    def apply_discount(price):
        discount_amount = price * discount_percentage / 100
        final_price = price - discount_amount
        return round(final_price, 2)

    # Return the inner function
    return apply_discount

# Create discount functions
discount_10 = create_discount(10)
discount_20 = create_discount(20)
discount_30 = create_discount(30)

# Accept product price from the user
price = float(input("Enter the product price: "))

# Display the discounted prices
print(f"Original Price : \u20b9{price:.2f}")
print(f"10% Discount   : \u20b9{discount_10(price):.2f}")
print(f"20% Discount   : \u20b9{discount_20(price):.2f}")
print(f"30% Discount   : \u20b9{discount_30(price):.2f}")
```

```python
#Assignment 91: Closure-Based Counter
# Outer function
def create_counter():
    count = 0

    # Inner function
    def counter():
        nonlocal count
        count += 1
        return count

    # Return the inner function
    return counter

# Create a counter
my_counter = create_counter()

# Call the returned function five times
print("Count:", my_counter())
print("Count:", my_counter())
print("Count:", my_counter())
print("Count:", my_counter())
print("Count:", my_counter())
```

```python
#Assignment 92: Closure-Based Bank Balance
# Outer function
def create_bank_account(initial_balance):

    balance = initial_balance

    # Deposit function
    def deposit(amount):
        nonlocal balance
        if amount > 0:
            balance += amount
            print(f"Deposited: \u20b9{amount:.2f}")
        else:
            print("Invalid deposit amount.")

    # Withdrawal function
    def withdraw(amount):
        nonlocal balance
        if amount <= 0:
            print("Invalid withdrawal amount.")
        elif amount > balance:
            print("Insufficient balance.")
        else:
            balance -= amount
            print(f"Withdrawn: \u20b9{amount:.2f}")

    # Balance enquiry function
    def check_balance():
        print(f"Current Balance: \u20b9{balance:.2f}")

    # Return all three inner functions
    return deposit, withdraw, check_balance


# Create a bank account with an initial balance
deposit, withdraw, check_balance = create_bank_account(5000)

# Perform operations
check_balance()

deposit(1500)
check_balance()

withdraw(2000)
check_balance()

withdraw(6000)   # Exceeds available balance
check_balance()
```

```python
#Assignment 93: Function Decorator
# Decorator function
def execution_logger(func):

    def wrapper(*args, **kwargs):
        print("Starting function execution...")

        # Call the original function
        result = func(*args, **kwargs)

        print("Function execution completed.")

        # Return the original function's result
        return result

    return wrapper


# Apply the decorator
@execution_logger
def add(a, b):
    return a + b


# Accept input from the user
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))

# Call the decorated function
sum_result = add(num1, num2)

# Display the result
print("Sum:", sum_result)
```

```python
#Assignment 100: Complete Functional Programming Project
from functools import reduce

# -------------------------------
# Global Variable (Global Scope)
# -------------------------------
company = "ABC Technologies"

# -------------------------------
# Sample Employee Data
# -------------------------------
employees = [
    {
        "id": 101,
        "name": "Ravi",
        "department": "IT",
        "salary": 60000,
        "performance_score": 88,
        "experience": 5
    },
    {
        "id": 102,
        "name": "Anitha",
        "department": "HR",
        "salary": 45000,
        "performance_score": 72,
        "experience": 4
    },
    {
        "id": 103,
        "name": "Kiran",
        "department": "IT",
        "salary": 75000,
        "performance_score": 94,
        "experience": 7
    },
    {
        "id": 104,
        "name": "Meena",
        "department": "Finance",
        "salary": 55000,
        "performance_score": 81,
        "experience": 6
    },
    {
        "id": 105,
        "name": "Arjun",
        "department": "IT",
        "salary": 40000,
        "performance_score": 65,
        "experience": 2
    }
]

# ------------------------------------------------
# Normal Function
# ------------------------------------------------
def display_employees(employee_list):
    print("\nEmployee Details")
    print("-" * 70)
    for emp in employee_list:
        print(
            f"{emp['id']} | {emp['name']:8} | "
            f"{emp['department']:8} | "
            f"Salary: {emp['salary']:6} | "
            f"Score: {emp['performance_score']} | "
            f"Experience: {emp['experience']} Years"
        )

# ------------------------------------------------
# map() + lambda
# Calculate Bonus
# ------------------------------------------------
def calculate_bonus():
    print("\nEmployee Bonus (10%)")
    print("-" * 35)

    bonuses = list(
        map(
            lambda emp: (
                emp["name"],
                round(emp["salary"] * 0.10, 2)
            ),
            employees
        )
    )

    for name, bonus in bonuses:
        print(f"{name:8} : \u20b9{bonus}")

# ------------------------------------------------
# filter() + lambda
# High Performers
# ------------------------------------------------
def high_performers():
    print("\nHigh Performers (Score >= 80)")
    print("-" * 35)

    result = list(
        filter(
            lambda emp: emp["performance_score"] >= 80,
            employees
        )
    )

    display_employees(result)

# ------------------------------------------------
# reduce()
# Total Salary
# ------------------------------------------------
def total_salary():
    total = reduce(
        lambda total, emp: total + emp["salary"],
        employees,
        0
    )

    print("\nTotal Salary Paid : \u20b9", total)

# ------------------------------------------------
# Nested Function
# ------------------------------------------------
def performance_report():

    print("\nPerformance Report")

    def grade(score):
        if score >= 90:
            return "Excellent"
        elif score >= 80:
            return "Very Good"
        elif score >= 70:
            return "Good"
        else:
            return "Needs Improvement"

    for emp in employees:
        print(
            emp["name"],
            "->",
            grade(emp["performance_score"])
        )

# ------------------------------------------------
# Closure
# ------------------------------------------------
def create_increment(percent):

    def increment(emp):
        new_salary = emp["salary"] * (1 + percent / 100)
        return round(new_salary, 2)

    return increment

# ------------------------------------------------
# LEGB Demonstration
# ------------------------------------------------
value = "Global Scope"

def legb_demo():

    value = "Enclosing Scope"

    def inner():

        value = "Local Scope"

        print("\nLEGB Demonstration")
        print("Local     :", value)

    inner()

    print("Enclosing :", value)

print("Global    :", value)

# ------------------------------------------------
# Main Program
# ------------------------------------------------

print("=" * 70)
print("EMPLOYEE PERFORMANCE ANALYSIS SYSTEM")
print("Company :", company)
print("=" * 70)

display_employees(employees)

calculate_bonus()

high_performers()

total_salary()

performance_report()

print("\nClosure Demonstration")
print("-" * 30)

increase_20 = create_increment(20)

for emp in employees:
    print(
        f"{emp['name']:8} Old Salary: \u20b9{emp['salary']} "
        f"New Salary: \u20b9{increase_20(emp)}"
    )

legb_demo()
```

```python
#Part 10: Nested Report Generator
# Sample Employee Data
employees = [
    {
        "id": 101,
        "name": "Ravi",
        "department": "IT",
        "salary": 60000,
        "performance_score": 88,
        "experience": 5
    },
    {
        "id": 102,
        "name": "Priya",
        "department": "HR",
        "salary": 55000,
        "performance_score": 92,
        "experience": 4
    },
    {
        "id": 103,
        "name": "Amit",
        "department": "Finance",
        "salary": 70000,
        "performance_score": 85,
        "experience": 6
    }
]

# Outer function
def generate_department_report(department):
    # Filter employees by department
    dept_employees = list(
        filter(lambda emp: emp["department"].lower() == department.lower(), employees)
    )

    # Nested function: Number of employees
    def employee_count():
        return len(dept_employees)

    # Nested function: Total salary
    def total_salary():
        return sum(emp["salary"] for emp in dept_employees)

    # Nested function: Average salary
    def average_salary():
        if employee_count() == 0:
            return 0
        return total_salary() / employee_count()

    # Nested function: Average performance score
    def average_performance():
        if employee_count() == 0:
            return 0
        total_score = sum(emp["performance_score"] for emp in dept_employees)
        return total_score / employee_count()

    # Return report as a dictionary
    return {
        "Department": department,
        "Number of Employees": employee_count(),
        "Total Salary": total_salary(),
        "Average Salary": average_salary(),
        "Average Performance Score": average_performance()
    }

# Example
report = generate_department_report("IT")

print("Department Report")
for key, value in report.items():
    print(f"{key}: {value}")
```

```python
#Part 12: Menu-Driven Program
from functools import reduce

# Sample Employee Data
employees = [
    {
        "id": 101,
        "name": "Ravi",
        "department": "IT",
        "salary": 60000,
        "performance_score": 88,
        "experience": 5
    },
    {
        "id": 102,
        "name": "Priya",
        "department": "HR",
        "salary": 55000,
        "performance_score": 92,
        "experience": 4
    },
    {
        "id": 103,
        "name": "Amit",
        "department": "Finance",
        "salary": 70000,
        "performance_score": 85,
        "experience": 6
    }
]

# 1. Display Employees
def display_employees():
    for emp in employees:
        print("-" * 40)
        for key, value in emp.items():
            print(f"{key.capitalize()}: {value}")
    print("-" * 40)

# 2. Display Employee Names
def display_employee_names():
    names = list(map(lambda emp: emp["name"], employees))
    print(names)

# 3. Display High Performers
def display_high_performers():
    high = list(filter(lambda emp: emp["performance_score"] >= 80, employees))
    for emp in high:
        print(emp)

# 4. Search by Department
def filter_by_department(department):
    result = list(filter(
        lambda emp: emp["department"].lower() == department.lower(),
        employees
    ))

    if result:
        for emp in result:
            print(emp)
    else:
        print("No employees found.")

# 5. Calculate Revised Salaries
def revised_salaries():
    revised = list(map(
        lambda emp: {
            "name": emp["name"],
            "old_salary": emp["salary"],
            "new_salary":
                emp["salary"] * 1.15 if emp["performance_score"] >= 90
                else emp["salary"] * 1.10 if emp["performance_score"] >= 80
                else emp["salary"] * 1.05 if emp["performance_score"] >= 70
                else emp["salary"]
        },
        employees
    ))

    for emp in revised:
        print(emp)

    return revised

# 6. Total Salary Expense
def total_salary_expense():
    revised = revised_salaries()

    current_total = reduce(
        lambda total, emp: total + emp["salary"],
        employees,
        0
    )

    revised_total = reduce(
        lambda total, emp: total + emp["new_salary"],
        revised,
        0
    )

    print("\nCurrent Total Salary :", current_total)
    print("Revised Total Salary :", revised_total)
    print("Additional Expense   :", revised_total - current_total)

# 7. Highest Paid Employee
def highest_paid_employee():
    highest = reduce(
        lambda e1, e2: e1 if e1["salary"] > e2["salary"] else e2,
        employees
    )

    print(highest)

# 8. Sort Employees
def sort_employees():
    print("\n1. Name")
    print("2. Salary")
    print("3. Performance Score")
    print("4. Experience")

    choice = input("Choose sorting option: ")

    if choice == "1":
        result = sorted(employees, key=lambda emp: emp["name"])
    elif choice == "2":
        result = sorted(employees, key=lambda emp: emp["salary"])
    elif choice == "3":
        result = sorted(employees, key=lambda emp: emp["performance_score"])
    elif choice == "4":
        result = sorted(employees, key=lambda emp: emp["experience"])
    else:
        print("Invalid choice.")
        return

    for emp in result:
        print(emp)

# 9. Department Report
def generate_department_report(department):
    dept = list(filter(
        lambda emp: emp["department"].lower() == department.lower(),
        employees
    ))

    if not dept:
        print("No employees found.")
        return

    report = {
        "Department": department,
        "Number of Employees": len(dept),
        "Total Salary": sum(emp["salary"] for emp in dept),
        "Average Salary": sum(emp["salary"] for emp in dept) / len(dept),
        "Average Performance":
            sum(emp["performance_score"] for emp in dept) / len(dept)
    }

    for key, value in report.items():
        print(f"{key}: {value}")

# 10. Menu
while True:
    print("\n===== Employee Performance Analysis System =====")
    print("1. Display all employees")
    print("2. Display employee names")
    print("3. Display high performers")
    print("4. Search by department")
    print("5. Calculate revised salaries")
    print("6. Calculate total salary expense")
    print("7. Find highest-paid employee")
    print("8. Sort employees")
    print("9. Generate department report")
    print("10. Exit")

    option = input("Enter your choice: ")

    if option == "1":
        display_employees()

    elif option == "2":
        display_employee_names()

    elif option == "3":
        display_high_performers()

    elif option == "4":
        department = input("Enter department: ")
        filter_by_department(department)

    elif option == "5":
        revised_salaries()

    elif option == "6":
        total_salary_expense()

    elif option == "7":
        highest_paid_employee()

    elif option == "8":
        sort_employees()

    elif option == "9":
        department = input("Enter department: ")
        generate_department_report(department)

    elif option == "10":
        print("Thank you for using the Employee Performance Analysis System.")
        break

    else:
        print("Invalid choice! Please try again.")
```
